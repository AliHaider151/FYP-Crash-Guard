```python
import cv2
import os
import yaml

# -------------------------------
# 1. Load dataset config from YAML
# -------------------------------
yaml_path = "assets/Accident-Detection-Dataset-Yolo/data.yaml"
with open(yaml_path, 'r') as f:
    data_cfg = yaml.safe_load(f)

images_folder = data_cfg['train']       # Train images path
labels_folder = images_folder.replace("images", "labels")  # Assuming labels folder is parallel
classes = data_cfg['names']

# -------------------------------
# 2. Helper: Draw YOLO boxes
# -------------------------------
def draw_yolo_box(img, label_path, classes):
    h, w = img.shape[:2]

    with open(label_path, "r") as f:
        for line in f.readlines():
            cls_id, x, y, bw, bh = map(float, line.split())
            cls_id = int(cls_id)

            x1 = int((x - bw/2) * w)
            y1 = int((y - bh/2) * h)
            x2 = int((x + bw/2) * w)
            y2 = int((y + bh/2) * h)

            cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
            cv2.putText(img, classes[cls_id], (x1, y1 - 8),
                        cv2.FONT_HERSHEY_SIMPLEX, 0.6, (0, 255, 0), 2)
    return img

# -------------------------------
# 3. Display first 100 images with progress
# -------------------------------
image_files = [f for f in os.listdir(images_folder) if f.endswith((".jpg", ".png"))]
total_images = min(len(image_files), 100)

for idx, img_file in enumerate(image_files[:100], 1):  # 1-based index
    img_path = os.path.join(images_folder, img_file)
    label_path = os.path.join(labels_folder, img_file.rsplit('.', 1)[0] + ".txt")

    img = cv2.imread(img_path)
    if img is None:
        print(f"Warning: could not read {img_path}")
        continue

    if os.path.exists(label_path):
        img = draw_yolo_box(img, label_path, classes)

    cv2.imshow("Dataset Preview", img)
    cv2.waitKey(100)  # Show each image for 100 ms

    # Print progress
    print(f"Loaded {idx}/{total_images} images")

cv2.destroyAllWindows()
```

```python
import cv2
import time
import os
import math
from collections import deque
from ultralytics import YOLO

# Load model
model = YOLO("best-train-1.pt")

# Load video
cap = cv2.VideoCapture("assets/Accident-Detection-Dataset-Yolo/videos/accident-video.mp4")
fps = int(cap.get(cv2.CAP_PROP_FPS))

# 10 sec buffer (before accident)
BUFFER_SECONDS = 10
buffer_size = fps * BUFFER_SECONDS
frame_buffer = deque(maxlen=buffer_size)

# After-accident duration
AFTER_SECONDS = 10

# Output folder
output_folder = "saved_clips"
os.makedirs(output_folder, exist_ok=True)
fourcc = cv2.VideoWriter_fourcc(*"mp4v")

# Accident classes
accident_classes = ["Accident", "mild", "moderate", "severe"]

# System state control
is_recording = False
recording_start_time = None
writer = None

def draw_boxes(frame, boxes, names):
    for box in boxes:
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        cls_id = int(box.cls[0])
        conf = float(box.conf[0])

        label = f"{names[cls_id]} {conf:.2f}"

        color = (0, 0, 255) if names[cls_id] in accident_classes else (0, 255, 0)

        cv2.rectangle(frame, (x1, y1), (x2, y2), color, 2)
        cv2.putText(frame, label, (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.6, color, 2)
    return frame

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Store frames in buffer
    frame_buffer.append(frame.copy())

    current_time = time.time()

    # ------------------------------------------------------
    # If recording clip (10sec BEFORE + 10sec AFTER accident)
    # ------------------------------------------------------
    if is_recording:
        # Write current frame
        writer.write(frame)

        # Check if 10 seconds after the accident have passed
        if current_time - recording_start_time >= AFTER_SECONDS:
            print("🎬 Clip finished. Saving & resetting...")
            writer.release()
            writer = None
            is_recording = False  # Ready for next accident
        else:
            # Continue but skip detection
            cv2.imshow("Live Detection", frame)
            if cv2.waitKey(1) & 0xFF == ord("q"):
                break
            continue

    # -----------------------------------------
    # DETECT ACCIDENTS (Only when not recording)
    # -----------------------------------------
    results = model.predict(frame, conf=0.25, show=False)
    boxes = results[0].boxes

    accident_detected = False

    for box in boxes:
        cls_name = model.names[int(box.cls[0])]
        if cls_name in accident_classes:
            accident_detected = True
            break

    # Draw boxes with class and confidence
    annotated_frame = draw_boxes(frame.copy(), boxes, model.names)

    # -----------------------------------------
    # START NEW ACCIDENT CLIP
    # -----------------------------------------
    if accident_detected and not is_recording:
        clip_id = int(time.time() * 1000)
        filename = os.path.join(output_folder, f"accident_{clip_id}.mp4")

        print(f"🔥 Accident Detected — Saving 20 sec clip: {filename}")

        writer = cv2.VideoWriter(filename, fourcc, fps,
                                 (frame.shape[1], frame.shape[0]))

        # Write previous 10 seconds of frames
        for bf in frame_buffer:
            writer.write(bf)

        # Start writing next 10 seconds
        recording_start_time = time.time()
        is_recording = True

    # Show annotated live video
    cv2.imshow("Live Detection", annotated_frame)
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

# Final cleanup
if writer:
    writer.release()

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import time
import os
import sqlite3
import hashlib
from collections import deque
from ultralytics import YOLO
import tkinter as tk
from tkinter import ttk, messagebox, filedialog
from PIL import Image, ImageTk
import threading


class Database:
    def __init__(self):
        self.conn = sqlite3.connect('accident_detection.db', check_same_thread=False)
        self.cursor = self.conn.cursor()
        self.create_tables()
    
    def create_tables(self):
        self.cursor.execute('''
            CREATE TABLE IF NOT EXISTS users (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT NOT NULL,
                email TEXT NOT NULL,
                phone TEXT NOT NULL,
                address TEXT NOT NULL,
                created_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
            )
        ''')
        self.conn.commit()
    
    def register_user(self, name, email, phone, address):
        try:
            self.cursor.execute(
                'INSERT INTO users (name, email, phone, address) VALUES (?, ?, ?, ?)',
                (name, email, phone, address)
            )
            self.conn.commit()
            return self.cursor.lastrowid
        except Exception as e:
            print(f"Error: {e}")
            return None


class AccidentDetector:
    def __init__(self, model_path="best-train-1.pt"):
        self.model = YOLO(model_path)
        self.accident_classes = ["Accident", "mild", "moderate", "severe"]
        self.BUFFER_SECONDS = 10
        self.AFTER_SECONDS = 10
        self.output_folder = "saved_clips"
        os.makedirs(self.output_folder, exist_ok=True)
        
    def detect_video(self, video_path, callback=None):
        cap = cv2.VideoCapture(video_path)
        fps = int(cap.get(cv2.CAP_PROP_FPS))
        
        buffer_size = fps * self.BUFFER_SECONDS
        frame_buffer = deque(maxlen=buffer_size)
        
        fourcc = cv2.VideoWriter_fourcc(*"mp4v")
        
        is_recording = False
        recording_start_time = None
        writer = None
        
        while True:
            ret, frame = cap.read()
            if not ret:
                break
            
            frame_buffer.append(frame.copy())
            current_time = time.time()
            
            if is_recording:
                writer.write(frame)
                
                if current_time - recording_start_time >= self.AFTER_SECONDS:
                    print("🎬 Clip finished. Saving & resetting...")
                    writer.release()
                    writer = None
                    is_recording = False
                else:
                    if callback:
                        callback(frame, "Recording...")
                    continue
            
            results = self.model.predict(frame, conf=0.25, show=False)
            annotated = results[0].plot()
            
            accident_detected = False
            
            for box in results[0].boxes:
                cls = self.model.names[int(box.cls)]
                if cls in self.accident_classes:
                    accident_detected = True
                    break
            
            if accident_detected and not is_recording:
                clip_id = int(time.time() * 1000)
                filename = os.path.join(self.output_folder, f"accident_{clip_id}.mp4")
                
                print(f"🔥 Accident Detected — Saving 20 sec clip: {filename}")
                
                writer = cv2.VideoWriter(filename, fourcc, fps,
                                       (frame.shape[1], frame.shape[0]))
                
                for bf in frame_buffer:
                    writer.write(bf)
                
                recording_start_time = time.time()
                is_recording = True
            
            status = "🔥 ACCIDENT DETECTED!" if accident_detected else "Monitoring..."
            if callback:
                callback(annotated, status)
        
        if writer:
            writer.release()
        
        cap.release()
        if callback:
            callback(None, "Video processing complete")


class ModernSignupWindow:
    def __init__(self, root, db, on_signup_success):
        self.root = root
        self.db = db
        self.on_signup_success = on_signup_success
        
        self.root.title("KAT King - Sign Up")
        self.root.geometry("1000x700")
        
        # Modern gradient background
        self.canvas = tk.Canvas(root, width=1000, height=700, highlightthickness=0)
        self.canvas.pack(fill="both", expand=True)
        
        # Create gradient background
        self.create_gradient_background()
        
        self.create_ui()
    
    def create_gradient_background(self):
        # Create a modern gradient effect
        colors = ['#667eea', '#764ba2']
        for i in range(700):
            ratio = i / 700
            r = int(102 + (118 - 102) * ratio)
            g = int(126 + (75 - 126) * ratio)
            b = int(234 + (162 - 234) * ratio)
            color = f'#{r:02x}{g:02x}{b:02x}'
            self.canvas.create_line(0, i, 1000, i, fill=color)
    
    def create_ui(self):
        # Main card frame
        card_frame = tk.Frame(self.canvas, bg="white", bd=0)
        card_frame.place(relx=0.5, rely=0.5, anchor="center", width=500, height=600)
        
        # Add shadow effect
        shadow = tk.Frame(self.canvas, bg="#d0d0d0", bd=0)
        shadow.place(relx=0.5, rely=0.505, anchor="center", width=510, height=610)
        shadow.lower(card_frame)
        
        # Logo/Icon area
        header_frame = tk.Frame(card_frame, bg="white", height=120)
        header_frame.pack(fill="x")
        header_frame.pack_propagate(False)
        
        # Icon circle
        icon_canvas = tk.Canvas(header_frame, width=80, height=80, bg="white", highlightthickness=0)
        icon_canvas.place(relx=0.5, rely=0.5, anchor="center")
        icon_canvas.create_oval(5, 5, 75, 75, fill="#667eea", outline="")
        icon_canvas.create_text(40, 40, text="🚗", font=("Arial", 30))
        
        # Title
        title_label = tk.Label(
            card_frame,
            text="KAT King Security",
            font=("Segoe UI", 26, "bold"),
            bg="white",
            fg="#2d3748"
        )
        title_label.pack(pady=(0, 5))
        
        subtitle_label = tk.Label(
            card_frame,
            text="AI-Powered Accident Detection System",
            font=("Segoe UI", 11),
            bg="white",
            fg="#718096"
        )
        subtitle_label.pack(pady=(0, 30))
        
        # Form container
        form_frame = tk.Frame(card_frame, bg="white")
        form_frame.pack(padx=50, fill="both", expand=True)
        
        # Full Name
        self.create_input_field(form_frame, "Full Name", "👤")
        self.name_entry = self.last_entry
        
        # Email
        self.create_input_field(form_frame, "Email Address", "📧")
        self.email_entry = self.last_entry
        
        # Phone
        self.create_input_field(form_frame, "Phone Number", "📱")
        self.phone_entry = self.last_entry
        
        # Address
        self.create_input_field(form_frame, "Full Address", "📍")
        self.address_entry = self.last_entry
        
        # Sign Up Button with gradient effect
        signup_btn = tk.Button(
            form_frame,
            text="Get Started →",
            font=("Segoe UI", 13, "bold"),
            bg="#667eea",
            fg="white",
            activebackground="#5568d3",
            activeforeground="white",
            bd=0,
            cursor="hand2",
            height=2,
            command=self.signup
        )
        signup_btn.pack(fill="x", pady=(25, 15))
        
        # Hover effect
        signup_btn.bind("<Enter>", lambda e: signup_btn.config(bg="#5568d3"))
        signup_btn.bind("<Leave>", lambda e: signup_btn.config(bg="#667eea"))
        
        # Footer
        footer_label = tk.Label(
            card_frame,
            text="Your data is secure and encrypted",
            font=("Segoe UI", 9),
            bg="white",
            fg="#a0aec0"
        )
        footer_label.pack(pady=(0, 20))
    
    def create_input_field(self, parent, placeholder, icon):
        field_frame = tk.Frame(parent, bg="white")
        field_frame.pack(fill="x", pady=(0, 15))
        
        # Container with border
        input_container = tk.Frame(field_frame, bg="#e2e8f0", bd=0)
        input_container.pack(fill="x")
        
        # Inner frame for padding
        inner_frame = tk.Frame(input_container, bg="white")
        inner_frame.pack(fill="x", padx=2, pady=2)
        
        # Icon
        icon_label = tk.Label(
            inner_frame,
            text=icon,
            font=("Segoe UI", 14),
            bg="white",
            fg="#667eea"
        )
        icon_label.pack(side="left", padx=(10, 5))
        
        # Entry
        entry = tk.Entry(
            inner_frame,
            font=("Segoe UI", 11),
            bg="white",
            fg="#2d3748",
            bd=0,
            insertbackground="#667eea"
        )
        entry.pack(side="left", fill="x", expand=True, padx=(5, 10), pady=12)
        entry.insert(0, placeholder)
        entry.config(fg="#a0aec0")
        
        # Placeholder behavior
        def on_focus_in(event):
            if entry.get() == placeholder:
                entry.delete(0, "end")
                entry.config(fg="#2d3748")
        
        def on_focus_out(event):
            if entry.get() == "":
                entry.insert(0, placeholder)
                entry.config(fg="#a0aec0")
        
        entry.bind("<FocusIn>", on_focus_in)
        entry.bind("<FocusOut>", on_focus_out)
        
        self.last_entry = entry
    
    def signup(self):
        name = self.name_entry.get().strip()
        email = self.email_entry.get().strip()
        phone = self.phone_entry.get().strip()
        address = self.address_entry.get().strip()
        
        # Validation
        if name == "Full Name" or not name:
            messagebox.showerror("Error", "Please enter your full name")
            return
        
        if email == "Email Address" or not email or "@" not in email:
            messagebox.showerror("Error", "Please enter a valid email address")
            return
        
        if phone == "Phone Number" or not phone:
            messagebox.showerror("Error", "Please enter your phone number")
            return
        
        if address == "Full Address" or not address:
            messagebox.showerror("Error", "Please enter your full address")
            return
        
        user_id = self.db.register_user(name, email, phone, address)
        
        if user_id:
            user_info = (user_id, name, email, phone, address)
            self.on_signup_success(user_info)
        else:
            messagebox.showerror("Error", "Registration failed. Please try again.")


class ModernMainApp:
    def __init__(self, root, user_info):
        self.root = root
        self.user_info = user_info
        self.detector = None
        self.video_thread = None
        self.is_processing = False
        
        self.root.title("KAT King - Accident Detection")
        self.root.geometry("1400x900")
        
        self.create_ui()
    
    def create_ui(self):
        # Main container with gradient
        main_container = tk.Frame(self.root, bg="#f7fafc")
        main_container.pack(fill="both", expand=True)
        
        # Top Navigation Bar
        nav_bar = tk.Frame(main_container, bg="white", height=80)
        nav_bar.pack(fill="x", side="top")
        nav_bar.pack_propagate(False)
        
        # Add subtle shadow
        shadow_line = tk.Frame(main_container, bg="#e2e8f0", height=2)
        shadow_line.pack(fill="x", side="top")
        
        # Logo section
        logo_frame = tk.Frame(nav_bar, bg="white")
        logo_frame.pack(side="left", padx=30)
        
        logo_canvas = tk.Canvas(logo_frame, width=50, height=50, bg="white", highlightthickness=0)
        logo_canvas.pack(side="left")
        logo_canvas.create_oval(5, 5, 45, 45, fill="#667eea", outline="")
        logo_canvas.create_text(25, 25, text="🚗", font=("Arial", 20))
        
        brand_label = tk.Label(
            logo_frame,
            text="KAT King",
            font=("Segoe UI", 18, "bold"),
            bg="white",
            fg="#2d3748"
        )
        brand_label.pack(side="left", padx=(15, 0))
        
        # User info section
        user_frame = tk.Frame(nav_bar, bg="white")
        user_frame.pack(side="right", padx=30)
        
        user_name_label = tk.Label(
            user_frame,
            text=f"👤 {self.user_info[1]}",
            font=("Segoe UI", 12, "bold"),
            bg="white",
            fg="#2d3748"
        )
        user_name_label.pack(anchor="e")
        
        user_loc_label = tk.Label(
            user_frame,
            text=f"📍 {self.user_info[4]}",
            font=("Segoe UI", 10),
            bg="white",
            fg="#718096"
        )
        user_loc_label.pack(anchor="e")
        
        # Content area
        content_area = tk.Frame(main_container, bg="#f7fafc")
        content_area.pack(fill="both", expand=True, padx=30, pady=30)
        
        # Left Panel - Control Card
        left_card = tk.Frame(content_area, bg="white", width=400)
        left_card.pack(side="left", fill="y", padx=(0, 15))
        left_card.pack_propagate(False)
        
        # Card header
        card_header = tk.Frame(left_card, bg="#667eea", height=100)
        card_header.pack(fill="x")
        card_header.pack_propagate(False)
        
        header_title = tk.Label(
            card_header,
            text="Detection Control",
            font=("Segoe UI", 20, "bold"),
            bg="#667eea",
            fg="white"
        )
        header_title.pack(pady=(25, 5))
        
        header_subtitle = tk.Label(
            card_header,
            text="Select and process videos",
            font=("Segoe UI", 11),
            bg="#667eea",
            fg="#e6e9ff"
        )
        header_subtitle.pack()
        
        # Card body
        card_body = tk.Frame(left_card, bg="white")
        card_body.pack(fill="both", expand=True, padx=25, pady=25)
        
        # Video selection section
        section_label = tk.Label(
            card_body,
            text="Video Source",
            font=("Segoe UI", 12, "bold"),
            bg="white",
            fg="#2d3748"
        )
        section_label.pack(anchor="w", pady=(0, 10))
        
        # Select video button
        self.select_btn = tk.Button(
            card_body,
            text="📁  Browse Video File",
            font=("Segoe UI", 12),
            bg="#f7fafc",
            fg="#2d3748",
            activebackground="#e2e8f0",
            bd=0,
            cursor="hand2",
            height=3,
            command=self.select_video
        )
        self.select_btn.pack(fill="x", pady=(0, 15))
        
        # Video info display
        self.video_info_frame = tk.Frame(card_body, bg="#edf2f7", bd=0)
        self.video_info_frame.pack(fill="x", pady=(0, 20))
        
        self.video_label = tk.Label(
            self.video_info_frame,
            text="No video selected",
            font=("Segoe UI", 10),
            bg="#edf2f7",
            fg="#718096",
            wraplength=320,
            justify="left"
        )
        self.video_label.pack(padx=15, pady=15)
        
        # Divider
        tk.Frame(card_body, bg="#e2e8f0", height=1).pack(fill="x", pady=20)
        
        # Control section
        control_label = tk.Label(
            card_body,
            text="Detection Controls",
            font=("Segoe UI", 12, "bold"),
            bg="white",
            fg="#2d3748"
        )
        control_label.pack(anchor="w", pady=(0, 10))
        
        # Start button
        self.start_btn = tk.Button(
            card_body,
            text="▶  Start Detection",
            font=("Segoe UI", 13, "bold"),
            bg="#48bb78",
            fg="white",
            activebackground="#38a169",
            activeforeground="white",
            bd=0,
            cursor="hand2",
            height=2,
            state="disabled",
            command=self.start_detection
        )
        self.start_btn.pack(fill="x", pady=(0, 10))
        
        # Stop button
        self.stop_btn = tk.Button(
            card_body,
            text="⏹  Stop Detection",
            font=("Segoe UI", 13, "bold"),
            bg="#f56565",
            fg="white",
            activebackground="#e53e3e",
            activeforeground="white",
            bd=0,
            cursor="hand2",
            height=2,
            state="disabled",
            command=self.stop_detection
        )
        self.stop_btn.pack(fill="x")
        
        # Status section
        tk.Frame(card_body, bg="#e2e8f0", height=1).pack(fill="x", pady=20)
        
        status_container = tk.Frame(card_body, bg="#edf2f7", bd=0)
        status_container.pack(fill="x")
        
        status_title = tk.Label(
            status_container,
            text="Status",
            font=("Segoe UI", 10, "bold"),
            bg="#edf2f7",
            fg="#4a5568"
        )
        status_title.pack(anchor="w", padx=15, pady=(15, 5))
        
        self.status_label = tk.Label(
            status_container,
            text="🟢 Ready",
            font=("Segoe UI", 11),
            bg="#edf2f7",
            fg="#38a169"
        )
        self.status_label.pack(anchor="w", padx=15, pady=(0, 15))
        
        # Right Panel - Video Display Card
        right_card = tk.Frame(content_area, bg="white")
        right_card.pack(side="right", fill="both", expand=True)
        
        # Video display header
        video_header = tk.Frame(right_card, bg="white", height=60)
        video_header.pack(fill="x")
        video_header.pack_propagate(False)
        
        video_title = tk.Label(
            video_header,
            text="Live Detection Feed",
            font=("Segoe UI", 16, "bold"),
            bg="white",
            fg="#2d3748"
        )
        video_title.pack(side="left", padx=25, pady=20)
        
        # Video canvas
        self.video_canvas = tk.Label(
            right_card,
            bg="#2d3748",
            fg="#a0aec0"
        )
        self.video_canvas.pack(fill="both", expand=True, padx=25, pady=(0, 25))
        
        # Placeholder content
        placeholder_frame = tk.Frame(self.video_canvas, bg="#2d3748")
        placeholder_frame.place(relx=0.5, rely=0.5, anchor="center")
        
        placeholder_icon = tk.Label(
            placeholder_frame,
            text="🎬",
            font=("Arial", 60),
            bg="#2d3748",
            fg="#4a5568"
        )
        placeholder_icon.pack()
        
        placeholder_text = tk.Label(
            placeholder_frame,
            text="Video feed will appear here\nSelect a video to begin",
            font=("Segoe UI", 14),
            bg="#2d3748",
            fg="#718096"
        )
        placeholder_text.pack(pady=10)
        
        self.selected_video = None
    
    def select_video(self):
        file_path = filedialog.askopenfilename(
            title="Select Video File",
            filetypes=[("Video Files", "*.mp4 *.avi *.mov"), ("All Files", "*.*")]
        )
        if file_path:
            self.selected_video = file_path
            filename = os.path.basename(file_path)
            self.video_label.config(
                text=f"✓ {filename}\n\n📂 {file_path}",
                fg="#2d3748"
            )
            self.video_info_frame.config(bg="#d4edda")
            self.video_label.config(bg="#d4edda")
            self.start_btn.config(state="normal")
    
    def start_detection(self):
        if not self.selected_video:
            messagebox.showerror("Error", "Please select a video first")
            return
        
        self.is_processing = True
        self.select_btn.config(state="disabled", bg="#e2e8f0")
        self.start_btn.config(state="disabled", bg="#9ae6b4")
        self.stop_btn.config(state="normal")
        self.status_label.config(text="🔴 Processing...", fg="#e53e3e")
        
        # Initialize detector
        if not self.detector:
            try:
                self.status_label.config(text="⏳ Loading Model...", fg="#f6ad55")
                self.root.update()
                self.detector = AccidentDetector()
            except Exception as e:
                messagebox.showerror("Error", f"Failed to load model: {str(e)}")
                self.reset_ui()
                return
        
        # Start detection in separate thread
        self.video_thread = threading.Thread(target=self.run_detection, daemon=True)
        self.video_thread.start()
    
    def run_detection(self):
        try:
            self.detector.detect_video(self.selected_video, self.update_frame)
        except Exception as e:
            print(f"Detection error: {e}")
            self.root.after(0, lambda: messagebox.showerror("Error", f"Detection failed: {str(e)}"))
        finally:
            self.root.after(0, self.reset_ui)
    
    def update_frame(self, frame, status):
        if frame is not None:
            # Convert frame to PhotoImage
            frame_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            
            # Get canvas size
            canvas_width = self.video_canvas.winfo_width()
            canvas_height = self.video_canvas.winfo_height()
            
            if canvas_width > 1 and canvas_height > 1:
                frame_resized = cv2.resize(frame_rgb, (canvas_width - 4, canvas_height - 4))
                img = Image.fromarray(frame_resized)
                photo = ImageTk.PhotoImage(image=img)
                
                self.video_canvas.config(image=photo, text="")
                self.video_canvas.image = photo
        
        if status:
            if "ACCIDENT" in status:
                self.status_label.config(text=f"🚨 {status}", fg="#e53e3e")
            else:
                self.status_label.config(text=f"🟢 {status}", fg="#38a169")
    
    def stop_detection(self):
        self.is_processing = False
        self.reset_ui()
    
    def reset_ui(self):
        self.select_btn.config(state="normal", bg="#f7fafc")
        self.start_btn.config(state="normal", bg="#48bb78")
        self.stop_btn.config(state="disabled", bg="#fc8181")
        self.status_label.config(text="🟢 Ready", fg="#38a169")


def main():
    db = Database()
    root = tk.Tk()
    
    def on_signup_success(user_info):
        # Clear signup window
        for widget in root.winfo_children():
            widget.destroy()
        
        # Create main app
        ModernMainApp(root, user_info)
    
    ModernSignupWindow(root, db, on_signup_success)
    root.mainloop()


if __name__ == "__main__":
    main()
```
